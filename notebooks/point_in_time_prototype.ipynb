# Point-in-time PAYEMS prototype

Reconstruct the value reported for August 2008 payroll employment as of September 12, 2008, compare it with the currently revised value for the same observation, and measure the revision. PAYEMS values are reported in thousands of people.

```python
import os
import sys
from pathlib import Path

import pandas as pd
from dotenv import find_dotenv, load_dotenv
from IPython.display import display

PROJECT_ROOT = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / "src" / "asof").is_dir()
)
sys.path.insert(0, str(PROJECT_ROOT / "src"))

from asof.fred import compare_vintages

load_dotenv(find_dotenv())
API_KEY = os.getenv("FRED_API_KEY")
assert API_KEY, "FRED_API_KEY was not loaded from .env"
```

```python
SERIES_ID = "PAYEMS"
OBSERVATION_DATE = "2008-08-01"
AS_OF_DATE = "2008-09-12"
```

```python
result = compare_vintages(
    series_id=SERIES_ID,
    observation_date=OBSERVATION_DATE,
    historical_as_of=AS_OF_DATE,
    api_key=API_KEY,
)

historical_value = result.historical.value
current_value = result.current.value
revision = result.revision
```

```python
comparison = pd.DataFrame(
    {
        "series_id": [SERIES_ID],
        "observation_date": [OBSERVATION_DATE],
        "historical_as_of": [AS_OF_DATE],
        "historical_value_thousands": [historical_value],
        "current_value_thousands": [current_value],
        "revision_thousands": [revision],
    }
)

display(comparison.style.format(precision=1, thousands=","))
print(
    f"August 2008 PAYEMS was revised by {revision:,.0f} thousand jobs: "
    f"from {historical_value:,.0f} thousand known on {AS_OF_DATE} "
    f"to {current_value:,.0f} thousand currently."
)
```
